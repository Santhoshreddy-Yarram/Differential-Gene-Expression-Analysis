```python
import pandas as pd
proteome_df=pd.read_excel("overlaps_dep's_Y_AL vs NRDR.xlsx")  
transcriptome_df=pd.read_excel("overlaps_deg's_Y_AL vs NRDR.xlsx")
```

```python
# Find overlapping genes
overlapping_genes = set(transcriptome_df['gene']).intersection(set(proteome_df['gene']))
print(f"total numer of overlapping genes:{len(overlapping_genes)}")

# Find DEGs and DEPs (padj < 0.05 and Qvalue < 0.05)
# degs: Differentially expressed genes from RNA-seq data with adjusted p-value < 0.05 (padj).
# deps: Differentially expressed proteins from proteomics with Q-value < 0.05.

degs = transcriptome_df[transcriptome_df['padj'] < 0.05]
deps = proteome_df[proteome_df['Qvalue'] < 0.05]

# Prints how many DEGs and DEPs are identified after filtering.
# The f"..." lets you insert that number directly into the string.
print(f"number of deg's:{(len(degs))}")
print(f"number of dep's:{(len(deps))}")

#degs.to_excel("overlaps_deg's.xlsx",index=False)
#deps.to_excel("overlaps_dep's.xlsx",index=False)

# Genes that are both DEG and DEP
deg_dep_genes = set(degs['gene']).intersection(set(deps['gene']))

print(f"number of deg's and dep's: {len(deg_dep_genes)}")

# Extract relevant data from both datasets
# Selects only necessary columns to keep the output clean and relevant:
# From degs: gene, log2FoldChange, and padj
# From deps: gene, AVG Log2 Ratio, and Qvalue

degs_filtered = degs[['gene', 'log2FoldChange', 'padj']]
deps_filtered = deps[['gene', 'AVG Log2 Ratio', 'Qvalue']]

# Merge DEGs and DEPs on gene
# Performs an inner join on the gene column to get only the overlapping DEGs and DEPs.
# The resulting deg_dep_merged DataFrame has info from both datasets side-by-side for each shared gene.

deg_dep_merged = pd.merge(degs_filtered, deps_filtered, on='gene', how='inner')

# Save the merged file
# deg_dep_merged.to_excel("overlapping_deg_dep_genes4.xlsx", index=False)
```
