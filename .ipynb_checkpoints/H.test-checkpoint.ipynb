```python
import pandas as pd
from scipy.stats import hypergeom

# Load overlapping genes file
overlap_df = pd.read_excel("Overlapping_Genes_O_AL vs DR.xlsx") 

# values
M = 2822  # Total unique genes
n = 1268     # Total DEPs
N = 183    # Total DEGs
x = 91   # Number of unique overlapping genes between degs and deps

# Perform hypergeometric test
p_value = hypergeom.sf(x - 1, M, n, N)

# Print results
print(f"Total DEPs (Proteome): {n}")
print(f"Total DEGs (Transcriptome): {N}")
print(f"Overlapping Genes: {x}")
print(f"Total Unique Genes (M): {M}")
print(f"Hypergeometric Test p-value: {p_value:.4f}")
```
