```python
import pandas as pd
import scipy.stats as stats

# Load the datasets
proteome = pd.read_excel("C:/Users/91918/Pro_O_DR_O_AL_DEGs.xlsx")
transcriptome = pd.read_excel("C:/Users/91918/rna_DEGs & log2fold of O_AL vs O_DR.xlsx")

```

```python
# Ensure column names match your dataset
#Renames columns to ensure both datasets have a consistent column name (gene) for proper merging.

proteome = proteome.rename(columns={'Genes': 'gene'})  # Adjust column names if necessary
transcriptome = transcriptome.rename(columns={'gene': 'gene'})
```

```python
# Step 1 - Find overlapping genes

#Uses pd.merge() to find overlapping genes between the proteome and transcriptome datasets.

#how='inner' ensures that only common genes in both datasets are kept.

overlap = pd.merge(proteome, transcriptome, on='gene', how='inner')
```

```python
# Get list sizes

total_proteins = len(proteome)  # Total number of DEPs
total_transcripts = len(transcriptome)  # Total number of DEGs
overlapping_genes = len(overlap)  # Overlapping genes
```

set(proteome['gene']) converts the gene list in the proteome dataset into a set, removing any duplicate genes.

set(transcriptome['gene']) does the same for the transcriptome dataset.

.union(set(transcriptome['gene'])) finds the total unique genes from both datasets, in gene coloumn

len(...) calculates the total number of unique genes across proteome and transcriptome datasets.

Why is set() Used?

If we directly take len(proteome['gene']) + len(transcriptome['gene']), we may double-count genes that appear in both datasets.

Using set() ensures that we only count unique genes.

```python
# Hypergeometric test for overlap significance

# M → Total number of unique genes in both datasets.

# n → Number of DEPs (proteins).

# N → Number of DEGs (genes).

# x → Number of overlapping genes.

# Hypergeometric Test (stats.hypergeom.sf()):
#Survival function (sf) computes the probability of observing x or more overlaps by random chance.
    
M = len(set(proteome['gene']).union(set(transcriptome['gene'])))  # Total unique genes
n = total_proteins  # DEPs detected
N = total_transcripts  # DEGs detected
x = overlapping_genes  # Observed overlap
p_hyper = stats.hypergeom.sf(x-1, M, n, N)
```

```python
# Save results
#overlap.to_excel("C:/Users/91918/Overlapping_Genes.xlsx", index=False)
```

```python
# Print results
print(f"Total DEPs (Proteome): {total_proteins}")
print(f"Total DEGs (Transcriptome): {total_transcripts}")
print(f"Overlapping Genes: {overlapping_genes}")
print(f"Hypergeometric Test p-value: {p_hyper:.4f}")

#:.4f formats the floating-point number (p_hyper) to 4 decimal places.

#Example:

# If p_hyper = 0.12345678, it will print as 0.1235.

# If p_hyper = 1.00000000, it will print as 1.0000.
```

The script identifies overlapping genes between transcriptomic (DEGs) and proteomic (DEPs) datasets.

It calculates the probability of this overlap occurring by random chance using the hypergeometric test.

The p-value interpretation helps determine whether the overlap is due to the drug effect or random chance.
