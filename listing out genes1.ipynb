```python
import pandas as pd
df=pd.read_excel("liver proteome2.xlsx")
```

```python
df
```

```python
filtered_df= df[df["Comparison (group1/group2)"] == "O_DR / O_AL"]
display(filtered_df)
```

```python
# drop_duplicates(subset="Genes") keeps only the first occurrence of each unique gene (along with its Q-value).

unique_genes = filtered_df[["Genes", "Qvalue","AVG Log2 Ratio"]].dropna().drop_duplicates(subset="Genes")
print(unique_genes)
```

```python
# Save to Excel 
unique_genes.to_excel("unique_genes.xlsx", index=False)
```

```python
num_unique_genes=len(unique_genes)
print(num_unique_genes)
```
