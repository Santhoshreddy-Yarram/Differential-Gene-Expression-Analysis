```python
import pandas as pd
import matplotlib.pyplot as plt

# Load and filter dataset
df = pd.read_excel("DEG'S data in transcriptome and proteome1.xlsx")
df = df[df["data type"] == "proteome"]

# Plot
plt.figure(figsize=(8, 5))
bars = plt.barh(df["comparison(group1/group2)"], df["deg's"], color="#000080", edgecolor="black")

# Add text inside bars
#loops through bars and values 

# bar: Represents each horizontal bar in the bar chart.

# i: Represents the number of DEGs associated with that bar.

# bars: This is a collection of bar objects created by plt.barh(). Each bar represents a different comparison in the proteome data.

# df["deg's"]: This is the column containing differentially expressed gene (DEG) counts for each comparison.

# zip(bars, df["deg's"]): This pairs each bar with its corresponding DEG count, so we can process them together in the loop.

# bar.get_width(): Gets the length of the bar (which represents the DEG count).

# - 20: Moves the text slightly to the left inside the bar for better visibility.

# bar.get_y(): Gets the Y-position of the bar on the graph.

# bar.get_height()/2: Moves the text vertically to the center of the bar.

# str(v) Converts the DEG count (v) into a string so it can be displayed as text.

# ha stands for Horizontal Alignment.
# ha="right" ensures the text sticks to the right side of the position we calculated earlier.

for bar, i in zip(bars, df["deg's"]):
    plt.text(bar.get_width()- 20, bar.get_y() + bar.get_height()/2,str(i), 
              ha="right", fontsize=12, color="white",)

# Labels and formatting
plt.xlabel("Number of DEGs ( P < 0.05)")
plt.ylabel("Comparison")
plt.title("DEGs in Proteome Data")
plt.gca().invert_yaxis()

# **Save the plot before showing**
plt.savefig("DEGs_Proteome.png", dpi=300, bbox_inches="tight")  

plt.show()
```
