```python
import pandas as pd
proteome=pd.read_excel("unique_genes_pro_O_AL vs NRDR.xlsx")  
transcriptome=pd.read_excel("filtered_unique_genes_O_AL vs NRDR.xlsx")
```

```python
# Find overlapping genes
# Merge on Gene to find overlapping genes

#Uses pd.merge() to find overlapping genes between the proteome and transcriptome datasets.

#how='inner' ensures that only common genes in both datasets are kept.

overlapping_genes = pd.merge(proteome,transcriptome, on="gene", how="inner")
```

```python
# Save the overlapping genes
overlapping_genes.to_excel("Overlapping_Genes.xlsx", index=False)
```

```python
# Print result
print(f"Number of overlapping genes: {len(overlapping_genes)}")
```
