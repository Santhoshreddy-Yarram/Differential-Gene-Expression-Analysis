```python
import pandas as pd 
from scipy.stats import binomtest
```

```python
#load the data
overlap = pd.read_excel("C:/Users/91918/Overlapping_Genes1 0_AL vs O_DR.xlsx")
```

| (bitwise OR) is used for combining multiple conditions in Pandas boolean filtering.

It means "either condition can be True" for a row to be selected.

Important Note: In Pandas, when combining conditions, you must use | (bitwise OR) instead of or, and & (bitwise AND) instead of and.



```python
# Identify co-directional and anti-directional changes
# Selects genes where transcriptome (log2FoldChange.1) and proteome (AVG Log2 Ratio) show opposite trends
# .copy() ensures the selected genes are stored in a new DataFrame (co_directional) to avoid warnings.
# same trend:

# Both positive → Upregulated in both transcriptome & proteome.

# Both negative → Downregulated in both transcriptome & proteome

# opposite trend:

# Positive in transcriptome but negative in proteome → Upregulated at RNA level, downregulated at protein level.

# Negative in transcriptome but positive in proteome → Downregulated at RNA level, upregulated at protein level.

# we used logiacl operators here (> & <) 
co_directional = overlap[
    (overlap['log2FoldChange.1'] > 0) & (overlap['AVG Log2 Ratio'] > 0) | 
    (overlap['log2FoldChange.1'] < 0) & (overlap['AVG Log2 Ratio'] < 0)
].copy()

anti_directional = overlap[
    (overlap['log2FoldChange.1'] > 0) & (overlap['AVG Log2 Ratio'] < 0) |
    (overlap['log2FoldChange.1'] < 0) & (overlap['AVG Log2 Ratio'] > 0)
].copy()
```

```python
# Add direction labels
# Adds a new column "Direction" to label each gene as either "Co-Directional" or "Anti-Directional".

co_directional["Direction"] = "Co-Directional"
anti_directional["Direction"] = "Anti-Directional"
```

```python
# Combine and save to a single Excel sheet
combined_df = pd.concat([co_directional, anti_directional])
#combined_df.to_excel("C:/Users/91918/Directional_Genes.xlsx", index=False)
```

Using binomial test (binomtest) to check:

Null Hypothesis (H₀): Co-directional and anti-directional genes occur equally (50% chance each).

Alternative Hypothesis (H₁): Co-directional genes occur significantly more often than anti-directional genes.

alternative='greater' → Tests if co-directional genes occur more frequently than expected by chance.



```python
# Perform binomial test
num_co, num_anti = len(co_directional), len(anti_directional)
p_value = binomtest(num_co, num_co + num_anti, p=0.5, alternative='greater').pvalue
```

```python
# Print results
# .4e is the scientific notation value,for the better readability
print(f"Co-Directional: {num_co}, Anti-Directional: {num_anti}, P-value: {p_value:.4e}")
```
