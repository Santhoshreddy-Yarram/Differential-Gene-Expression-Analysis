```python
import pandas as pd 
df=pd.read_excel("liver proteome1.xlsx")
```

```python
df
```

```python
# Filter for the "O_DR / O_AL" comparison with P-value < 0.05
df_filtered = df[(df["Comparison (group1/group2)"] == "O_DR / O_AL") & (df["Qvalue"] < 0.05)]

# Display the first few rows
display(df_filtered.head())
```

```python
# Extract only the gene names that are differentially expressed
de_genes = df_filtered["Genes"].tolist()

# Display first 10 genes
print(de_genes[:10])
```

```python
print(f"Number of differentially expressed genes (Q < 0.05) in O_DR / O_AL: {df_filtered.shape[0]}")
```

```python


# Save filtered data to an Excel file
output_file = "O_DR_O_AL_DEGs.xlsx"
df_filtered[["Genes","AVG Log2 Ratio","Qvalue"]].to_excel(output_file, index=False, engine='openpyxl')

print(f"Differentially expressed genes saved to {output_file}")
```
