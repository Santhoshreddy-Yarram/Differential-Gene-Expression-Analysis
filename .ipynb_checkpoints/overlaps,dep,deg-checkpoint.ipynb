```python
import pandas as pd
proteome=pd.read_excel("unique_genes_pro_O_AL vs DR.xlsx")  
transcriptome=pd.read_excel("filtered_unique_genes_O_AL vs DR.xlsx")
```

```python
# Step 1: Find overlapping genes
overlap_genes = set(transcriptome['gene']).intersection(set(proteome['gene']))
print(f"Overlapping genes: {len(overlap_genes)}")
```

```python
# Convert the set to a DataFrame
#overlap_genes = pd.DataFrame(list(overlap_genes), columns=['gene',])
```

```python

#save results
#overlap_genes.to_excel("overlapping_genes.xlsx", index= False)
```

```python
# Step 2: Filter DEGs in each dataset
transcriptome_deg = transcriptome[(transcriptome['padj'] < 0.05) & (transcriptome['gene'].isin(overlap_genes))]
proteome_deg = proteome[(proteome['Qvalue'] < 0.05) & (proteome['gene'].isin(overlap_genes))]
num_degs= len(transcriptome_deg)
num_deps=len(proteome_deg)
print(num_degs)
print(num_deps)
# Step 3: Find common DEGs in both datasets
common_deg_genes = set(transcriptome_deg['gene']).intersection(set(proteome_deg['gene']))
```

```python
print(f"Common differentially expressed genes in both datasets: {len(common_deg_genes)}")
```
