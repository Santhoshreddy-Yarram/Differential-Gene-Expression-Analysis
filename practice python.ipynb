# dictonaries

```python
#dictionaries in python are mutable and we can chnage and add values

dict={
    "name":"santhosh",
    "age": 24,
    "study" : "bioinformatics",
    "subjects":["python","R","pandas","numpy"],
    "grade" : 85.5,
    "is_adult":True
}
print(dict)
```

```python
# if you wnat to chnage the value and  assign the new value
dict["name"]="devi"
print(dict)
```

```python
#if you wan to add another key into dictinaries
dict["surname"] = "Yarram"
print(dict)
```

```python
#crating null (empty) dictionaries adding values inside the dictionaries
null_dict={} #it is  called as null dictionary
print(null_dict)
```

```python
null_dict["name"]="savithri"
print(null_dict)
```

```python
# nested dictionaries
student = {
    "name":"nani",
    "subjects":{        #nested dictionary means dictionary inside dictionary
       "phy":90,
        "math":88,
        "chem":91}
}
print(student)
```

```python
print(student["subjects"])
```

```python
# if you wnat to extract the values from nested dictionaries
print(student["subjects"]["chem"])
```

 # methods

```python
#if you want to print the all keys form the dictionary
#dict.keys() method
print(student.keys())
```

```python
# the above results are which dict_keys 
# we can convert them into list or tuple if we want
# so we have to use type cast for that

print(list(student.keys()))
```

```python
 print(len(student))
```

```python
# values method
# it will give all the values of keys inside dictionary
print(student.values())
```

```python
print (list(student.values()))
```

```python
#dict.items() method 
# return all(key,val) pairs as tuples
print(student.items())
```

```python
print(list(student.items()))
```

```python
 #if you want to access the indiviuals of this tuple
pairs=list(student.items())
print(pairs[0])
```

```python
#dict.get() method 
#return the key according to the value
print(student.get("subjects"))
#print(student["subjects"])  # both will give results but .get() method is prefer more to avoid the error
```

```python
# dict.update() method
# it will store new values inside the dictionaries
student.update({"city":"hyderabad"})
```

# Sets


```python
# sets are un-ordered and only stores unique values ignore duplicates
# elements in sets are immuatable we can't change them but sets are mutable
# and we can't store lists and dictionaries inside the sets
#nums = {1.2,3,4} it's a set
```

```python
collection={1,2,3,4,}
print(collection)
print(type(collection))
```

```python
collection={1, 2, 2, "hello", "santhosh", "santhosh"}
print(collection)
```

```python
# now the above result's proved that sets are unordered 
```

```python
len(collection) # only print unique collections
```

```python
# if you want to create an empty set
#the syntax for creating an empty syntax is
collection = set()
print(type(collection))
```

# methods

```python
# adding values inside the sets
#set.add() method
collection = set()    # it will create an empty set

collection.add(1)     # adding values inside the empty set
collection.add(2)
collection.add("santhosh")
print(collection)
```

```python
# set.remove()method
collection.remove(1)
print(collection)
```

```python
# set.clear() method # it will remove everything from the set
```

```python
# set.pop()   # it will some values randomly
collection.pop()
```

```python
print(collection)
```

```python
# set.union() method
# in this method it will combine 2 sets and make a list of unique values from these sets
set1={1,2,3}
set2={2,3,4}
print(set1.union(set2)) # it will count duplicate values only one time
```

```python
#set.intersection() method 
# in this method it will combine 2 sets and make a list of common values from these sets.
set1={1,2,3}
set2={2,3,4}
print(set1.intersection(set2))   #it will only count the common values 
```
