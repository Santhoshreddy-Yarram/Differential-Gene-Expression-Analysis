```python
# break and continue statements in while loop
i=0
while i<5:
    print(i)
    if (i == 4):
        break
    i+=1
    
```

```python
# continue
#it will skip the current iteration or turn of the loop and it continue
i=0
while i<=5:
    if (i == 3):
         i+=1
         continue
    print(i)
    i+=1
```

```python
#range function in for loop
#range(start,stop,step)
#start = 0 by default
#stop stops before a specified number
#step = increments by 1 default

#example
for i in range(10):
    print(i)
#it will print from 0 to 9


```

```python
for i in range(1,10):# (start and stop)
    print(i)
#it will print from 1 to 9
```

```python
for i in range(1,10,2):# (start,stop,step)
    print(i)
#it will print from 1 to 9 with step of 2
```
