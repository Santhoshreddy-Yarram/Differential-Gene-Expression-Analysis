```python
import pandas as pd 
from scipy.stats import binomtest
```

```python
#load the data
overlap = pd.read_excel("C:/Users/91918/Overlapping_Genes_O_AL vs DR.xlsx")
```

```python
co_directional = overlap[
    (overlap['log2FoldChange.1'] > 0) & (overlap['AVG Log2 Ratio'] > 0) | 
    (overlap['log2FoldChange.1'] < 0) & (overlap['AVG Log2 Ratio'] < 0)
].copy()

anti_directional = overlap[
    (overlap['log2FoldChange.1'] > 0) & (overlap['AVG Log2 Ratio'] < 0) |
    (overlap['log2FoldChange.1'] < 0) & (overlap['AVG Log2 Ratio'] > 0)
].copy()
```

```python
co_directional["Direction"] = "Co-Directional"
anti_directional["Direction"] = "Anti-Directional"
```

```python
# Combine and save to a single Excel sheet
combined_df = pd.concat([co_directional, anti_directional])
#combined_df.to_excel("C:/Users/91918/Directional_Genes.xlsx", index=False)
```

```python
# Perform binomial test
num_co, num_anti = len(co_directional), len(anti_directional)
p_value = binomtest(num_co, num_co + num_anti, p=0.5, alternative='greater').pvalue
```

```python
# Print results
# .4e is the scientific notation value,for the better readability
print(f"Co-Directional: {num_co}, Anti-Directional: {num_anti}, P-value: {p_value:.4e}")
```

```python
#load the data
import pandas as pd
overlap4 = pd.read_excel("C:/Users/91918/Overlapping_Genes_Y_AL vs NRDR.xlsx")
```

```python
#This line filters genes where:
#Both transcriptome and proteome show upregulation (log2FC > 0 and AVG Log2 Ratio > 0) or
#Both show downregulation (log2FC < 0 and AVG Log2 Ratio < 0)
#These are co-directional genes → expression goes the same way in RNA and protein.


co_directional4 = overlap4[
    (overlap4['log2FoldChange.1'] > 0) & (overlap4['AVG Log2 Ratio'] > 0) | 
    (overlap4['log2FoldChange.1'] < 0) & (overlap4['AVG Log2 Ratio'] < 0)
].copy()

# Filtering genes where:

#Transcriptome is up, but proteome is down, or

Transcriptome is down, but proteome is up

These are anti-directional genes → expression goes in opposite directions between RNA and protein.



anti_directional4 = overlap4[
    (overlap4['log2FoldChange.1'] > 0) & (overlap4['AVG Log2 Ratio'] < 0) |
    (overlap4['log2FoldChange.1'] < 0) & (overlap4['AVG Log2 Ratio'] > 0)
].copy()
```

```python
co_directional4["Direction"] = "Co-Directional4"
anti_directional4["Direction"] = "Anti-Directional4"
```

```python
# Print results
# .4e is the scientific notation value,for the better readability
print(f"Co-Directiona: {num_co4}, Anti-Directional: {num_anti4}, P-value: {p_value4:.4e}")
```
